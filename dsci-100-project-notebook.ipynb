Visualization 1: This boxplot shows the seed of teams at each tournament finish. It relates to the research question because if the seed predicts finish, teams that go further in the tournament should have lower seeds than teams that lose earlier. This plot shows whether the seed distributions separate evenly across finishes.

```r
wmm_plot <- wmm_tidy |>
            mutate(tourney_finish = factor(tourney_finish, levels = c("opening_round_loss", "first_round_loss", "second_round_loss",
                                              "top_16_loss", "top_8_loss", "top_4_loss", "top_2_loss", "champ")))

seed_finish_boxplot <- wmm_plot |>
                ggplot(aes(x = tourney_finish, y = seed)) +
                geom_boxplot() +
                labs(x = "Tournament finish", y = "Seed (1 = best)",
                title = "Seed by Tournament Finish") +
                theme(axis.text.x = element_text(angle = 45, hjust = 1))
seed_finish_boxplot
```

Visualization 2: This is a bar chart of the average seed for each tournament finish. This is important to the research question it summarizes the relationship in a single number for each finish. If the average seed drops steadily as teams advance further, that means seed is a useful predictor of how a team goes.

```r
mean_seed_plot <- wmm_plot |>
                group_by(tourney_finish) |>
                summarize(mean_seed = mean(seed)) |>
                ggplot(aes(x = tourney_finish, y = mean_seed)) +
                geom_col() +
                labs(x = "Tournament finish", y = "Mean seed (1 = best)", title = "Mean Seed by Tournament Finish") +
                theme(axis.text.x = element_text(angle = 45, hjust = 1))
mean_seed_plot
```
